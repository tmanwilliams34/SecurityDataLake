```python
import pandas as pd
import os
from glob import glob
import gc
import numpy as np
```

Gather all files together using glob

```python
parquet_files = glob("../analytics/*.parquet")
```

For gathered files, change back to csv and save into computer.

```python
for file in parquet_files:
    print(f"Loading {file}...")
    df = pd.read_parquet(file)

    output_name = os.path.basename(file).replace(".parquet", ".csv")
    output_path = f"../analytics_csv/{output_name}"

    print(f"Converting {file} to csv...")
    df.to_csv(output_path, index=False)

    print(f"Finished {file}")
    print(f"Saved to: {output_path}")
```
