Import all necessary functions and tools

```python
import pandas as pd
from glob import glob
import numpy as np
import os
import gc
```

Check all files to see if there are any str that need to be converted into int.

```python
for file in glob("../analytics/*.parquet"):
    df = pd.read_parquet(file)

    print(
        file,
        "| Dst Port:", df["Dst Port"].dtype,
        "| Protocol:", df["Protocol"].dtype
    )
```

Create code for taking numeric columns, and int columns, then using for loops to replace the specified columns type

```python
numeric_cols = [
    "Dst Port",
    "Protocol",
    "Flow Duration",
    "Flow Byts/s",
    "Flow Pkts/s",
    "Tot Fwd Pkts",
    "Tot Bwd Pkts"
]

int_cols = [
    "Dst Port",
    "Protocol",
    "Flow Duration",
    "Tot Fwd Pkts",
    "Tot Bwd Pkts"
]

for file in glob("../analytics/*.parquet"):
    print(f"Fixing {file}...")

    df = pd.read_parquet(file)

    df = df[df["Label"] != "Label"]
    df["Timestamp"] = pd.to_datetime(df["Timestamp"], errors="coerce")

    for col in numeric_cols:
        df[col] = pd.to_numeric(df[col], errors="coerce")

    df = df.replace([np.inf, -np.inf], np.nan)

    df = df.dropna(subset=[
        "Dst Port",
        "Protocol",
        "Timestamp",
        "Label",
        "Flow Duration",
        "Tot Fwd Pkts",
        "Tot Bwd Pkts"
    ])

    for col in int_cols:
        df[col] = df[col].astype("int64")

    df.to_parquet(file, index=False)

    print("Saved:", file)
    print(df.dtypes[["Dst Port", "Protocol", "Timestamp", "Label"]])
    print()

    del df
    gc.collect()
```

Double check that all types were changed correctly.

```python
for file in glob("../analytics/*.parquet"):
    df = pd.read_parquet(file)

    print(
        file,
        "| Dst Port:", df["Dst Port"].dtype,
        "| Protocol:", df["Protocol"].dtype
    )
```
