Import all necessary tools for working with data and my system.

```python
import pandas as pd
import os
from glob import glob
import gc
import numpy as np
```

Create dataset cleaner, then round up and gather files. Run for loop to upload, clean, and save each file individually.

```python
def clean_dataset(df):
    df = df.drop_duplicates()
    df = df[df["Label"] != "Label"]
    df = df.dropna(subset=["Flow Byts/s"])
    df["Timestamp"] = pd.to_datetime(df["Timestamp"], errors="coerce")
    df = df.dropna(subset=["Timestamp"])
    return df

csv_files = glob("raw/*.csv")   # use this if notebook is inside raw/

for file in csv_files:

    print(f"Loading {file}...")
    df = pd.read_csv(file, low_memory=False)

    print(f"Cleaning {file}...")
    
    clean_df = clean_dataset(df)

    output_name = os.path.basename(file).replace(".csv", "-clean.parquet")
    output_path = f"cleaned/{output_name}"
    
    print(f"Writing {file}...")
    clean_df.to_parquet(output_path, index=False)


    print(f"Finished {file}")
    print("Original:", df.shape)
    print("Cleaned:", clean_df.shape)
    print("Saved to:", output_path)
    print()

    del df
    del clean_df
    gc.collect()
```
